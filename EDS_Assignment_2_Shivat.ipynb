```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv("Book Reviews.csv")

# Display basic info
df.info()
df.head()
```

### 1. Display the total number of reviews per book.

```python
df['Book'].value_counts()
```

### 2. Find the number of missing reviews.

```python
df['Review'].isna().sum()
```

### 3. Drop rows with missing reviews.

```python
df_cleaned = df.dropna(subset=['Review'])
df_cleaned.shape
```

### 4. Convert 'Review Date' to datetime format.

```python
df_cleaned['Review Date'] = pd.to_datetime(df_cleaned['Review Date'], errors='coerce')
df_cleaned['Review Date'].head()
```

### 5. Find the earliest and latest review date.

```python
df_cleaned['Review Date'].min(), df_cleaned['Review Date'].max()
```

### 6. Count the number of reviews per year.

```python
df_cleaned['Year'] = df_cleaned['Review Date'].dt.year
df_cleaned['Year'].value_counts().sort_index()
```

### 7. Get top 5 most reviewed books.

```python
df_cleaned['Book'].value_counts().head()
```

### 8. Count how many reviews contain the word 'love'.

```python
df_cleaned['Review'].str.contains('love', case=False, na=False).sum()
```

### 9. Add a column with the length of each review.

```python
df_cleaned['Review Length'] = df_cleaned['Review'].str.len()
df_cleaned[['Review', 'Review Length']].head()
```

### 10. Find average review length per book.

```python
df_cleaned.groupby('Book')['Review Length'].mean().sort_values(ascending=False).head()
```

### 11. Show standard deviation of review lengths using NumPy.

```python
np.std(df_cleaned['Review Length'].dropna())
```

### 12. Create a column for word count in each review.

```python
df_cleaned['Word Count'] = df_cleaned['Review'].apply(lambda x: len(str(x).split()))
df_cleaned[['Review', 'Word Count']].head()
```

### 13. Find the review with the maximum word count.

```python
df_cleaned.loc[df_cleaned['Word Count'].idxmax()]
```

### 14. Plot the distribution of review lengths (optional for visual).

```python
# import matplotlib.pyplot as plt\n# df_cleaned['Review Length'].hist(bins=50)
```

### 15. Find books with the highest average word count per review.

```python
df_cleaned.groupby('Book')['Word Count'].mean().sort_values(ascending=False).head()
```

### 16. Create a column showing if the review is long (>100 words).

```python
df_cleaned['Long Review'] = df_cleaned['Word Count'] > 100
df_cleaned[['Review', 'Long Review']].head()
```

### 17. Count number of long reviews per book.

```python
df_cleaned.groupby('Book')['Long Review'].sum().sort_values(ascending=False).head()
```

### 18. Find correlation between review length and word count.

```python
df_cleaned[['Review Length', 'Word Count']].corr()
```

### 19. Remove duplicate reviews (if any).

```python
df_cleaned = df_cleaned.drop_duplicates(subset='Review')
df_cleaned.shape
```

### 20. Reset the index of the cleaned dataset.

```python
df_cleaned.reset_index(drop=True, inplace=True)
df_cleaned.head()
```
